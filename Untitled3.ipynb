```python

import random

# Customer Service Chatbot

print("=" * 55)
print("       WELCOME TO CUSTOMER SERVICE CHATBOT")
print("=" * 55)

print("Bot: Hello! Welcome to our customer support service.")
print("Bot: How can I help you today?")
print("Bot: Type 'bye' to exit.\n")


# Knowledge Base: Questions and Answers

knowledge_base = {

    "greeting": {
        "keywords": ["hello", "hi", "hey", "good morning", "good evening"],
        "responses": [
            "Hello! How can I help you today?",
            "Hi there! Welcome to customer support.",
            "Good to see you! What can I do for you?"
        ]
    },

    "order_status": {
        "keywords": ["order status", "track order", "where is my order",
                     "order tracking", "delivery status"],
        "responses": [
            "You can track your order using the tracking ID sent to your email.",
            "Please visit the My Orders section to check your order status.",
            "Your order status can be checked through our order tracking page."
        ]
    },

    "payment": {
        "keywords": ["payment", "paid", "payment failed", "transaction",
                     "payment methods", "upi", "credit card"],
        "responses": [
            "We accept UPI, debit cards, credit cards and net banking.",
            "If your payment failed, please check your bank account.",
            "For a deducted amount with a failed order, please contact support."
        ]
    },

    "delivery": {
        "keywords": ["delivery", "shipping", "late delivery",
                     "delivery time", "when will it arrive"],
        "responses": [
            "Standard delivery usually takes 3-7 business days.",
            "You can check the estimated delivery date in My Orders.",
            "We apologise for the delay. Please check your tracking details."
        ]
    },

    "return": {
        "keywords": ["return", "return product", "send back",
                     "exchange", "replace product"],
        "responses": [
            "You can request a return within 7 days of delivery.",
            "Visit the My Orders section and select the return option.",
            "Please ensure that the product meets our return conditions."
        ]
    },

    "refund": {
        "keywords": ["refund", "money back", "refund status",
                     "refund time", "money returned"],
        "responses": [
            "Refunds are generally processed within 5-7 business days after approval.",
            "You can check your refund status in the My Orders section.",
            "The refunded amount will be credited to your original payment method."
        ]
    },

    "cancellation": {
        "keywords": ["cancel order", "cancellation", "cancel my order",
                     "stop order"],
        "responses": [
            "You can cancel an order before it is shipped.",
            "Go to My Orders and select the cancellation option.",
            "Orders that have already been shipped may not be cancellable."
        ]
    },

    "account": {
        "keywords": ["account", "login", "sign in", "password",
                     "forgot password", "register"],
        "responses": [
            "You can log in using your registered email address.",
            "Use the Forgot Password option to reset your password.",
            "You can create a new account using the Sign Up option."
        ]
    },

    "product": {
        "keywords": ["product", "products", "availability",
                     "stock", "price", "details"],
        "responses": [
            "You can explore available products on our website.",
            "Product prices and details are available on each product page.",
            "Please check the product page for current stock availability."
        ]
    },

    "discount": {
        "keywords": ["discount", "offer", "coupon", "promo code",
                     "voucher", "sale"],
        "responses": [
            "Check our offers section for the latest discounts.",
            "You can apply a valid coupon code during checkout.",
            "Some promotional offers have minimum purchase requirements."
        ]
    },

    "complaint": {
        "keywords": ["complaint", "bad service", "poor service",
                     "problem", "issue", "unhappy"],
        "responses": [
            "We are sorry for the inconvenience. Please explain your issue.",
            "We value your feedback and will try to resolve your problem.",
            "Please share more details so we can assist you."
        ]
    },

    "contact": {
        "keywords": ["contact", "customer care", "support",
                     "phone number", "email", "human agent"],
        "responses": [
            "You can contact our support team through the Contact Us page.",
            "Our customer support team is available to assist you.",
            "Please visit the Help Centre for additional assistance."
        ]
    },

    "working_hours": {
        "keywords": ["working hours", "opening hours", "available time",
                     "support timing"],
        "responses": [
            "Our customer support is available Monday to Saturday, 9 AM to 6 PM.",
            "You can contact our support team during business hours."
        ]
    },

    "thanks": {
        "keywords": ["thank you", "thanks", "thank"],
        "responses": [
            "You're welcome! Happy to help.",
            "Glad I could help you!",
            "Thank you for contacting customer support."
        ]
    },

    "goodbye": {
        "keywords": ["bye", "goodbye", "see you", "exit"],
        "responses": [
            "Thank you for contacting us. Have a great day!",
            "Goodbye! Feel free to contact us again.",
            "Take care! Have a wonderful day."
        ]
    }
}


# Function to find the chatbot response

def chatbot_response(user_input):

    user_input = user_input.lower().strip()

    # Check every category
    for category, data in knowledge_base.items():

        for keyword in data["keywords"]:

            if keyword in user_input:

                return random.choice(data["responses"])

    # Default response if no keyword matches

    return (
        "Sorry, I couldn't understand your question. "
        "Could you please explain it differently?"
    )


# Main chatbot loop

while True:

    user_input = input("You: ")

    if user_input.lower().strip() in ["bye", "goodbye", "exit", "quit"]:
        print("Bot: Thank you for using our customer service chatbot!")
        break

    response = chatbot_response(user_input)

    print("Bot:", response)
```
